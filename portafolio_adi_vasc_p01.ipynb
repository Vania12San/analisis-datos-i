Para mi proyecto de Mercadotecnia Estratégica en Google Colab, voy a analizar la base de datos de Profeco para entender cómo interactúan el pricing, la competencia y la segmentación en el mercado real. Trabajar con este conjunto de registros me permitirá evaluar la dispersión de precios de productos clave como el aceite entre distintas cadenas comerciales y estados, identificando las estrategias de fijación de precios de cada retail (desde Everyday Low Price hasta High-Low Pricing).

```r
library(tidyverse)
```

```r
library(readr)
```

```r
datos <- read_csv("11-2025_01.csv")
```

```r
datos
```

```r
head(datos)
```

```r
glimpse(datos)
```

```r
datos <- datos |>
  select(
    producto = producto,
    presentacion = presentacion,
    marca = marca,
    categoria = categoria,
    precio = precio,
    fecha = fecha_registro,
    cadena = cadena_comercial,
    giro = giro,
    tienda = nombre_comercial,
    estado = estado,
    municipio = municipio
    )
```

```r
glimpse(datos)
```

```r
cadenas_top <- c("Walmart", "Soriana", "Bodega Aurrera", "Chedraui", "Mega Soriana")
```

```r
datos_filtrados <- datos %>%
  filter(producto == "Aceite", cadena %in% cadenas_top)
```

```r
ggplot(data = datos_filtrados, mapping = aes(x = cadena, y = precio)) +
  geom_point(aes(color = cadena)) +
  geom_smooth(method = "lm")
```
